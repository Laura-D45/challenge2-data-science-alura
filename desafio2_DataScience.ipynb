#📌 Extracción

```python
import pandas as pd

#Importacion de los datos directamente desde la api
api = 'https://raw.githubusercontent.com/ingridcristh/challenge2-data-science-LATAM/refs/heads/main/TelecomX_Data.json'
#Conversion de la data en un dataframe
df = pd.read_json(api)
df.head()
```

```python
#Ultimos registros
df.tail()
```

#🔧 Transformación

```python
#Exploracion de columnas del dataset y visualizacion de las columnas con sus respectivos tipos de dato
df.info()
# 7267 registros
```

```python
#dimensionamiento del dataset
df.shape
```

```python
#Estadistica basica y valores unicos
df.describe()
```

```python
# ver valores nulos
df.isnull().sum()
```

```python
# Valores unicos en la columna Churn ya que es una que no tiene datos en un formato json
df['Churn'].unique()
```

```python
#Los datos de Churn se colocaran en minuscula
df['Churn'] = df['Churn'].str.lower()
df['Churn'].head()
```

```python
# estandarización o transformación
df['Churn'] = df['Churn'].replace({'true': 'yes', 'false': 'no'})
df.head()
```

#📊 Carga y análisis

```python
df.describe()
```

```python
#Cantidad de valores que pertenecen a cada uno de los atributos  de churn
df['Churn'].value_counts()
```

```python
df['Churn'].unique()
#Quiere decir que ese espacio vacio
```

```python
import matplotlib.pyplot as plt

df['Churn'].value_counts().plot(kind='bar')

plt.title('Distribución de Churn')
plt.xlabel('Churn')
plt.ylabel('Clientes')

plt.show()
```

```python
# Normalizar el data frame columnas que vienen en formato json
customer_df = pd.json_normalize(df['customer'])
phone_df = pd.json_normalize(df['phone'])
internet_df = pd.json_normalize(df['internet'])
account_df = pd.json_normalize(df['account'])
```

```python
# Unir todo normalizado en un df
df_final = pd.concat(
    [
        df[['customerID','Churn']],
        customer_df,
        phone_df,
        internet_df,
        account_df
    ],
    # unir todo horizontalmente
    axis=1
)
df_final.head()
```

```python
# prueba de que ya se puede ingresar al contenido de columnas que estaban en formato json
df_final['gender'].value_counts()
```

```python
# Identificar las variables numéricas y de resto son object o str
df_final.select_dtypes(include=['int64','float64']).columns
```

```python
#Validacion de que ya no son json
df_final.info()
```

```python
#Grafica de porcentaje de Churn por Tipo de Contrato
churn_contract = pd.crosstab(df_final['Contract'], df_final['Churn'], normalize='index') * 100
import matplotlib.pyplot as plt

churn_contract.plot(kind='bar', stacked=True)

plt.title('Tasa de Churn según tipo de contrato')
plt.ylabel('Porcentaje de clientes')
plt.xlabel('Tipo de contrato')

plt.legend(title='Churn')

plt.show()
```

# Introducción

El problema de evasión de clientes (Churn) es uno de los principales desafíos para las empresas que ofrecen servicios recurrentes. Esto ocurre cuando un cliente decide cancelar o dejar de utilizar el servicio ofrecido.

El objetivo de este análisis es explorar y comprender los factores que influyen en la evasión de clientes utilizando un dataset que contiene información demográfica, características de servicio contratado y datos de facturación.

A través de técnicas de análisis exploratorio de datos (EDA), se busca identificar patrones y relaciones entre las variables que permitan entender mejor el comportamiento de los clientes y detectar posibles factores asociados con la cancelación del servicio.

Los resultados obtenidos pueden servir como base para el desarrollo de estrategias orientadas a mejorar la retención de clientes y reducir la tasa de churn.

# Análisis Exploratorio de Datos

El análisis exploratorio se realizó con el objetivo de identificar patrones y relaciones entre las variables que puedan estar asociadas con la evasión de clientes.

Primero se analizó la distribución general de la variable **Churn**, observando la proporción de clientes que cancelaron el servicio frente a aquellos que permanecieron.

Posteriormente, se exploró la relación entre la evasión y la variable categórica como:

• Género del cliente  


Este análisis permitió identificar segmentos de clientes con mayor tendencia a cancelar el servicio.

También se analizaron variables numéricas como:

• Tiempo de permanencia del cliente (tenure)  
• Cargos mensuales (MonthlyCharges)  
• Total gastado por el cliente (TotalCharges)  

Las visualizaciones obtenidas ayudan a comprender mejor el perfil de los clientes con mayor probabilidad de evasión.

---

# Conclusiones e Insights

A partir del análisis exploratorio de los datos se identificaron varios patrones relevantes relacionados con la evasión de clientes.

Uno de los principales hallazgos es que los clientes con menor tiempo de permanencia en el servicio presentan una mayor probabilidad de cancelar su suscripción.

También se observó que los clientes con contratos mensuales tienden a cancelar con mayor frecuencia en comparación con aquellos que poseen contratos a largo plazo.

---

# Recomendaciones

Con base en el análisis, se pueden proponer algunas acciones estratégicas para reducir la evasión de clientes:

• Incentivar contratos a largo plazo mediante descuentos o beneficios adicionales para los clientes.

• Analizar la estructura de precios para evaluar si los cargos mensuales influyen en la decisión de cancelación.

• Mejorar la experiencia del cliente en los segmentos donde se observa mayor churn.

• Desarrollar modelos predictivos de machine learning que permitan identificar clientes con alto riesgo de evasión y actuar de forma preventiva.
